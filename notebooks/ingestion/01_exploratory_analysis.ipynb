```python
import pandas as pd
import glob
from loguru import logger

def load_latest(name: str) -> pd.DataFrame:
    """Carrega o arquivo Parquet mais recente para um dataset."""
    pattern = f"../../data/raw/{name}_*.parquet"
    files = sorted(glob.glob(pattern))
    if not files:
        raise FileNotFoundError(f"Nenhum arquivo encontrado para: {pattern}")
    latest = files[-1]
    logger.info(f"Carregando: {latest}")
    return pd.read_parquet(latest)

customers = load_latest("customers")
products  = load_latest("products")
orders    = load_latest("orders")
items     = load_latest("order_items")
```

```python
print("=== CLIENTES ===")
print(f"Total: {len(customers):,}")
print(f"Ativos: {customers['is_active'].sum():,}")
print(f"Por segmento:\n{customers['segment'].value_counts()}")
print(f"Nulos:\n{customers.isnull().sum()}")
```

```python
print("\n=== PEDIDOS ===")
print(f"Total: {len(orders):,}")
print(f"Por status:\n{orders['status'].value_counts()}")
print(f"Ticket médio: R$ {orders['total_amount'].mean():.2f}")
print(f"Receita total: R$ {orders['total_amount'].sum():,.2f}")
```

```python
print("\n=== QUALIDADE ===")
pct_null = customers['loyalty_score'].isna().mean() * 100
print(f"Clientes sem loyalty_score: {pct_null:.1f}%")
pct_coupon = orders['coupon_code'].notna().mean() * 100
print(f"Pedidos com cupom: {pct_coupon:.1f}%")
```
