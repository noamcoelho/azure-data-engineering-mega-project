```python
%%sql
SELECT customer_id, name, segment, city, state
FROM customers
WHERE segment = 'Gold'
LIMIT 10;
```

```python
%%sql
SELECT segment, COUNT(*) as total_clientes
FROM customers
GROUP BY segment
ORDER BY total_clientes DESC;
```

```python
%%sql
SELECT category, AVG(sale_price) as preco_medio
FROM products
GROUP BY category
HAVING AVG(sale_price) > 200
ORDER BY preco_medio DESC;
```

```python
%%sql
SELECT 
    o.order_id,
    o.order_date,
    o.total_amount,
    c.name,
    c.segment
FROM orders o
INNER JOIN customers c ON o.customer_id = c.customer_id
LIMIT 10;
```

```python
%%sql
-- Encontra clientes que NUNCA fizeram um pedido
SELECT c.customer_id, c.name, c.segment
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL;
```

```python
%%sql
SELECT 
    o.order_id,
    c.name as cliente,
    p.name as produto,
    oi.quantity,
    oi.subtotal
FROM orders o
INNER JOIN customers c    ON o.customer_id  = c.customer_id
INNER JOIN order_items oi ON o.order_id     = oi.order_id
INNER JOIN products p     ON oi.product_id  = p.product_id
LIMIT 10;
```

```python
%%sql
SELECT 
    c.customer_id,
    c.name,
    SUM(o.total_amount) as total_gasto,
    RANK() OVER (ORDER BY SUM(o.total_amount) DESC) as ranking
FROM customers c
INNER JOIN orders o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.name
ORDER BY ranking
LIMIT 10;
```

```python
%%sql
-- Top 3 produtos mais vendidos POR CATEGORIA
SELECT * FROM (
    SELECT 
        p.category,
        p.name,
        SUM(oi.quantity) as total_vendido,
        ROW_NUMBER() OVER (
            PARTITION BY p.category 
            ORDER BY SUM(oi.quantity) DESC
        ) as posicao
    FROM products p
    INNER JOIN order_items oi ON p.product_id = oi.product_id
    GROUP BY p.category, p.name
) ranked
WHERE posicao <= 3;
```

```python
%%sql
-- Receita diária e variação em relação ao dia anterior
SELECT 
    DATE(order_date) as dia,
    SUM(total_amount) as receita_dia,
    LAG(SUM(total_amount)) OVER (ORDER BY DATE(order_date)) as receita_dia_anterior
FROM orders
GROUP BY DATE(order_date)
ORDER BY dia
LIMIT 15;
```

```python
%%sql
WITH receita_por_cliente AS (
    SELECT 
        customer_id,
        SUM(total_amount) as total_gasto,
        COUNT(*) as total_pedidos
    FROM orders
    WHERE status = 'delivered'
    GROUP BY customer_id
),
clientes_vip AS (
    SELECT customer_id, total_gasto
    FROM receita_por_cliente
    WHERE total_gasto > 5000
)
SELECT 
    c.name,
    c.segment,
    v.total_gasto
FROM clientes_vip v
INNER JOIN customers c ON v.customer_id = c.customer_id
ORDER BY v.total_gasto DESC
LIMIT 10;
```

```python
%%sql
SELECT 
    customer_id,
    name,
    CASE 
        WHEN segment = 'Platinum' THEN 'VIP'
        WHEN segment = 'Gold'     THEN 'Premium'
        ELSE 'Regular'
    END as categoria_cliente,
    is_active
FROM customers
LIMIT 10;
```

```python
query = """
SELECT 
    p.category,
    SUM(oi.subtotal) as receita_total,
    COUNT(DISTINCT oi.order_id) as total_pedidos,
    ROUND(AVG(oi.subtotal), 2) as ticket_medio_item
FROM products p
INNER JOIN order_items oi ON p.product_id = oi.product_id
GROUP BY p.category
ORDER BY receita_total DESC;
"""
df = pd.read_sql(query, conn)
print("=== Receita total por categoria ===")
df
```

```python
query = """
SELECT 
    o.order_id,
    o.customer_id,
    o.order_date,
    o.total_amount,
    ROW_NUMBER() OVER (
        PARTITION BY o.customer_id 
        ORDER BY o.order_date
    ) as posicao_compra
FROM orders o
ORDER BY o.customer_id, posicao_compra
LIMIT 20;
"""
df = pd.read_sql(query, conn)
print("=== Sequência de compras por cliente ===")
df
```

```python
query = """
WITH ultima_compra AS (
    SELECT 
        customer_id,
        MAX(order_date) as data_ultimo_pedido,
        COUNT(*) as total_pedidos
    FROM orders
    GROUP BY customer_id
),
data_referencia AS (
    SELECT MAX(order_date) as data_max FROM orders
),
clientes_inativos AS (
    SELECT 
        u.customer_id,
        u.data_ultimo_pedido,
        u.total_pedidos,
        CAST(
            JULIANDAY((SELECT data_max FROM data_referencia)) - 
            JULIANDAY(u.data_ultimo_pedido) 
        AS INTEGER) as dias_sem_compra
    FROM ultima_compra u
    WHERE CAST(
        JULIANDAY((SELECT data_max FROM data_referencia)) - 
        JULIANDAY(u.data_ultimo_pedido) 
    AS INTEGER) > 90
)
SELECT 
    c.customer_id,
    c.name,
    c.segment,
    c.email,
    ci.data_ultimo_pedido,
    ci.dias_sem_compra,
    ci.total_pedidos
FROM clientes_inativos ci
INNER JOIN customers c ON ci.customer_id = c.customer_id
WHERE c.is_active = 1
ORDER BY ci.dias_sem_compra DESC;
"""
df = pd.read_sql(query, conn)
print(f"=== Clientes em risco de churn: {len(df)} clientes ===")
df.head(15)
```

```python
query = """
WITH estatisticas AS (
    SELECT 
        AVG(total_amount) as media,
        AVG(total_amount * total_amount) - AVG(total_amount) * AVG(total_amount) as variancia
    FROM orders
),
limites AS (
    SELECT 
        media,
        media + 3 * SQRT(variancia) as limite_superior
    FROM estatisticas
)
SELECT 
    o.order_id,
    o.customer_id,
    o.order_date,
    o.total_amount,
    ROUND(l.media, 2) as media_geral,
    ROUND(l.limite_superior, 2) as limite_3_desvios
FROM orders o, limites l
WHERE o.total_amount > l.limite_superior
ORDER BY o.total_amount DESC;
"""
df = pd.read_sql(query, conn)
print(f"=== Pedidos outliers: {len(df)} registros ===")
df
```

```python
query = """
WITH receita_mensal AS (
    SELECT 
        STRFTIME('%Y-%m', order_date) as mes,
        SUM(total_amount)             as receita_total,
        COUNT(*)                      as total_pedidos,
        ROUND(AVG(total_amount), 2)   as ticket_medio
    FROM orders
    WHERE status != 'cancelled'
    GROUP BY STRFTIME('%Y-%m', order_date)
),
resumo_com_variacao AS (
    SELECT 
        mes,
        receita_total,
        total_pedidos,
        ticket_medio,
        LAG(receita_total) OVER (ORDER BY mes) as receita_mes_anterior,
        ROUND(
            (receita_total - LAG(receita_total) OVER (ORDER BY mes)) 
            / LAG(receita_total) OVER (ORDER BY mes) * 100
        , 1) as variacao_pct
    FROM receita_mensal
)
SELECT 
    mes,
    receita_total,
    total_pedidos,
    ticket_medio,
    receita_mes_anterior,
    variacao_pct,
    CASE 
        WHEN variacao_pct > 0  THEN ' Alta'
        WHEN variacao_pct < 0  THEN ' Queda'
        WHEN variacao_pct = 0  THEN ' Estável'
        ELSE '— Primeiro mês'
    END as tendencia
FROM resumo_com_variacao
ORDER BY mes;
"""
df = pd.read_sql(query, conn)
print("=== Resumo Executivo Mensal — DataCommerce ===")
df
```
